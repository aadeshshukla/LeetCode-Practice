## Introduction to Scikit-learn

`scikit-learn` (often referred to as `sklearn`) is a free software machine learning library for the Python programming language. It features various classification, regression and clustering algorithms including support vector machines, random forests, gradient boosting, k-means and DBSCAN, and is designed to interoperate with the Python numerical and scientific libraries NumPy and SciPy.

Let's demonstrate a basic classification task using a synthetic dataset.

```python
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate a synthetic dataset
X, y = make_blobs(n_samples=100, centers=3, random_state=42, cluster_std=1.0)

print("Shape of features (X):", X.shape)
print("Shape of labels (y):", y.shape)

# Visualize the synthetic dataset
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', s=50, edgecolors='k')
plt.title('Synthetic Dataset for Classification')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.colorbar(label='Class')
plt.show()

# 2. Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"\nTraining data points: {len(X_train)}")
print(f"Testing data points: {len(X_test)}")

# 3. Choose a model (K-Nearest Neighbors Classifier)
knn = KNeighborsClassifier(n_neighbors=5)

# 4. Train the model
knn.fit(X_train, y_train)

# 5. Make predictions on the test set
y_pred = knn.predict(X_test)

# 6. Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"\nModel Accuracy: {accuracy:.2f}")

# Visualize decision boundary (optional, for better understanding)
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                     np.arange(y_min, y_max, 0.1))
Z = knn.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.figure(figsize=(8, 6))
plt.contourf(xx, yy, Z, alpha=0.8, cmap='viridis')
plt.scatter(X_test[:, 0], X_test[:, 1], c=y_test, cmap='viridis', s=50, edgecolors='k')
plt.title('K-NN Classifier Decision Boundary with Test Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

In this `scikit-learn` example, we demonstrated a typical machine learning workflow:

1.  **Generate Data**: We used `make_blobs` to create a simple, separable dataset with features (`X`) and corresponding labels (`y`).
2.  **Split Data**: `train_test_split` is crucial for evaluating model performance on unseen data. It divides the dataset into training and testing sets.
3.  **Choose a Model**: We selected `KNeighborsClassifier` as our model, a simple yet effective classification algorithm.
4.  **Train the Model**: The `fit()` method learns patterns from the training data.
5.  **Make Predictions**: `predict()` uses the trained model to classify new, unseen data (our test set).
6.  **Evaluate the Model**: `accuracy_score` measures how well our model performed on the test set. The decision boundary visualization helps understand how the model separates different classes.

## Regression Algorithms: Linear Regression

Regression algorithms are used to predict continuous values, unlike classification which predicts discrete categories. Linear Regression is one of the simplest and most widely used algorithms for this task. It models the relationship between a dependent variable and one or more independent variables by fitting a linear equation to the observed data.

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate a synthetic dataset for regression
# We generate a dataset with 100 samples, 1 feature, and some noise
X_reg, y_reg = make_regression(n_samples=100, n_features=1, noise=20, random_state=42)

print("Shape of features (X_reg):", X_reg.shape)
print("Shape of target (y_reg):", y_reg.shape)

# Visualize the synthetic regression dataset
plt.figure(figsize=(8, 6))
plt.scatter(X_reg, y_reg, s=50, alpha=0.7, label='Data points')
plt.title('Synthetic Dataset for Linear Regression')
plt.xlabel('Feature')
plt.ylabel('Target Variable')
plt.legend()
plt.grid(True)
plt.show()

# 2. Split the dataset into training and testing sets
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.3, random_state=42)

print(f"\nTraining data points for regression: {len(X_train_reg)}")
print(f"Testing data points for regression: {len(X_test_reg)}")

# 3. Choose a model (Linear Regression)
linear_model = LinearRegression()

# 4. Train the model
linear_model.fit(X_train_reg, y_train_reg)

# 5. Make predictions on the test set
y_pred_reg = linear_model.predict(X_test_reg)

# 6. Evaluate the model
mse = mean_squared_error(y_test_reg, y_pred_reg)
r2 = r2_score(y_test_reg, y_pred_reg)

print(f"\nMean Squared Error (MSE): {mse:.2f}")
print(f"R-squared (R2) Score: {r2:.2f}")

# Visualize the regression line
plt.figure(figsize=(8, 6))
plt.scatter(X_test_reg, y_test_reg, s=50, alpha=0.7, label='Actual Values')
plt.plot(X_test_reg, y_pred_reg, color='red', linewidth=2, label='Regression Line')
plt.title('Linear Regression: Actual vs. Predicted')
plt.xlabel('Feature')
plt.ylabel('Target Variable')
plt.legend()
plt.grid(True)
plt.show()
```

In this Linear Regression example, we followed a similar machine learning workflow:

1.  **Generate Data**: `make_regression` created a dataset where `y_reg` is a linear function of `X_reg` plus some random noise.
2.  **Split Data**: We separated the data into training and testing sets to evaluate the model's generalization ability.
3.  **Choose and Train Model**: `LinearRegression` was instantiated and trained using the training data (`fit()`).
4.  **Make Predictions**: The trained model predicted `y_pred_reg` for the test features (`predict()`).
5.  **Evaluate Model**: We used two common metrics:
    *   **Mean Squared Error (MSE)**: The average of the squared differences between predicted and actual values. Lower is better.
    *   **R-squared (R2) Score**: Represents the proportion of the variance in the dependent variable that is predictable from the independent variable(s). A value closer to 1 indicates a better fit.

We also visualized the scatter plot of the data along with the fitted regression line to see how well it captures the trend.

## Clustering Algorithms: K-Means

Clustering is a type of unsupervised learning that aims to find patterns and group similar data points together without any prior knowledge of labels. K-Means is a popular and relatively simple clustering algorithm that partitions `n` observations into `k` clusters, where each observation belongs to the cluster with the nearest mean (centroid), serving as a prototype of the cluster.

```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate a synthetic dataset for clustering
# We'll create data that naturally forms a few clusters
X_cluster, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

print("Shape of features (X_cluster):", X_cluster.shape)

# Visualize the original (unknown) clusters
plt.figure(figsize=(8, 6))
plt.scatter(X_cluster[:, 0], X_cluster[:, 1], s=50, cmap='viridis', label='Data points (True Clusters)')
plt.title('Synthetic Dataset for Clustering (True Clusters)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()

# 2. Apply K-Means clustering
# Let's assume we know there are 4 clusters (k=4)
kmeans = KMeans(n_clusters=4, random_state=0, n_init=10) # n_init is set to suppress a future warning
kmeans.fit(X_cluster)

# Get the cluster assignments (labels) and centroids
y_kmeans = kmeans.predict(X_cluster)
centers = kmeans.cluster_centers_

print("\nPredicted cluster labels for first 10 data points:", y_kmeans[:10])
print("Cluster centroids:\n", centers)

# 3. Visualize the K-Means clusters
plt.figure(figsize=(8, 6))
plt.scatter(X_cluster[:, 0], X_cluster[:, 1], c=y_kmeans, s=50, cmap='viridis', label='Clustered Data')
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.7, marker='X', label='Cluster Centroids')
plt.title('K-Means Clustering Results')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```

In this K-Means clustering example, we:

1.  **Generated Data**: Used `make_blobs` to create a dataset with clear, separable clusters, but we treat the true labels (`y_true`) as unknown to simulate an unsupervised scenario.
2.  **Applied K-Means**: Instantiated `KMeans` with `n_clusters=4` (assuming we know the optimal number of clusters for demonstration) and fitted it to our data using `fit()`.
3.  **Visualized Results**: Plotted the data points colored by their assigned cluster (`y_kmeans`) and highlighted the calculated cluster centroids (the red 'X' markers).

K-Means aims to minimize the within-cluster sum of squares. The algorithm iteratively assigns data points to the nearest centroid and then recomputes the centroids based on the new cluster assignments until convergence.

## Classification Algorithms: Support Vector Machines (SVM)

Support Vector Machines (SVMs) are a set of supervised learning models with associated learning algorithms that analyze data for classification and regression analysis. SVMs are particularly effective in high-dimensional spaces and cases where the number of dimensions is greater than the number of samples. They work by finding an optimal hyperplane that best separates the classes in the feature space.

```python
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.datasets import make_classification
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate a synthetic dataset for classification
# We'll create a dataset with two features and two classes, suitable for SVM
X_svm, y_svm = make_classification(n_samples=100, n_features=2, n_informative=2, n_redundant=0, n_clusters_per_class=1, random_state=42)

print("Shape of features (X_svm):", X_svm.shape)
print("Shape of labels (y_svm):", y_svm.shape)

# Visualize the synthetic dataset
plt.figure(figsize=(8, 6))
plt.scatter(X_svm[:, 0], X_svm[:, 1], c=y_svm, cmap='bwr', s=50, edgecolors='k')
plt.title('Synthetic Dataset for SVM Classification')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.colorbar(label='Class')
plt.grid(True)
plt.show()

# 2. Split the dataset into training and testing sets
X_train_svm, X_test_svm, y_train_svm, y_test_svm = train_test_split(X_svm, y_svm, test_size=0.3, random_state=42)

print(f"\nTraining data points for SVM: {len(X_train_svm)}")
print(f"Testing data points for SVM: {len(X_test_svm)}")

# 3. Choose a model (Support Vector Classifier - SVC)
# We'll use a linear kernel for simplicity to start
svm_model = SVC(kernel='linear', random_state=42)

# 4. Train the model
svm_model.fit(X_train_svm, y_train_svm)

# 5. Make predictions on the test set
y_pred_svm = svm_model.predict(X_test_svm)

# 6. Evaluate the model
accuracy_svm = accuracy_score(y_test_svm, y_pred_svm)
print(f"\nSVM Model Accuracy: {accuracy_svm:.2f}")

# Visualize decision boundary
x_min, x_max = X_svm[:, 0].min() - 1, X_svm[:, 0].max() + 1
y_min, y_max = X_svm[:, 1].min() - 1, X_svm[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                     np.arange(y_min, y_max, 0.02))

Z = svm_model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.figure(figsize=(8, 6))
plt.contourf(xx, yy, Z, cmap='bwr', alpha=0.5)
plt.scatter(X_svm[:, 0], X_svm[:, 1], c=y_svm, cmap='bwr', s=50, edgecolors='k')
plt.scatter(X_test_svm[:, 0], X_test_svm[:, 1], s=80, facecolors='none', edgecolors='green', label='Test Data')
plt.title('SVM Classifier Decision Boundary')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)
plt.show()
```

## Introduction to Neural Networks: The "Neuron"

A **Neural Network** is a series of algorithms that endeavors to recognize underlying relationships in a set of data through a process that mimics the way the human brain operates. At the core of every neural network is the concept of a **neuron** (also called a perceptron).

### What is a Neuron?

A single artificial neuron is a mathematical function that takes one or more inputs, applies a transformation to them, and produces an output. It's inspired by its biological counterpart:

*   **Inputs (Dendrites)**: Each input is a numerical value, often representing a feature from your data. Each input is associated with a **weight**, which determines its importance.
*   **Weighted Sum (Cell Body)**: The neuron calculates a weighted sum of its inputs, meaning each input is multiplied by its corresponding weight, and all these products are added together. A **bias** term is also added to this sum.
*   **Activation Function (Axon)**: The weighted sum (plus bias) is then passed through an **activation function**. This function introduces non-linearity to the neuron's output, allowing the network to learn complex patterns. Common activation functions include ReLU, Sigmoid, and Tanh.
*   **Output (Axon Terminal)**: The result of the activation function is the neuron's output, which can then be fed as input to other neurons in the network.

### How Neurons Form a Network?

Neurons are organized into **layers**:

1.  **Input Layer**: Receives the raw data features. There's one neuron per feature.
2.  **Hidden Layer(s)**: One or more layers of neurons between the input and output layers. These layers learn complex representations of the data.
3.  **Output Layer**: Produces the final output of the network, which could be a classification (e.g., probability of a class) or a regression value.

The connections between neurons have adjustable weights, which are learned during the training process. The network learns by adjusting these weights to minimize the difference between its predictions and the actual target values.

Let's build a simple Feedforward Neural Network using `tensorflow.keras` for a classification task.

```python
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate a synthetic dataset for classification
# We'll create a dataset with two features and two classes for simplicity
X_nn, y_nn = make_classification(n_samples=1000, n_features=2, n_informative=2, n_redundant=0, n_clusters_per_class=1, random_state=42)

print("Shape of features (X_nn):", X_nn.shape)
print("Shape of labels (y_nn):", y_nn.shape)

# Visualize the synthetic dataset
plt.figure(figsize=(8, 6))
plt.scatter(X_nn[:, 0], X_nn[:, 1], c=y_nn, cmap='bwr', s=50, edgecolors='k')
plt.title('Synthetic Dataset for Neural Network Classification')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.colorbar(label='Class')
plt.grid(True)
plt.show()

# 2. Split the dataset into training and testing sets
X_train_nn, X_test_nn, y_train_nn, y_test_nn = train_test_split(X_nn, y_nn, test_size=0.3, random_state=42)

# 3. Preprocess the data: Scaling features
# Neural networks often perform better when input features are scaled
scaler = StandardScaler()
X_train_nn_scaled = scaler.fit_transform(X_train_nn)
X_test_nn_scaled = scaler.transform(X_test_nn)

print(f"\nTraining data points for Neural Network: {len(X_train_nn_scaled)}")
print(f"Testing data points for Neural Network: {len(X_test_nn_scaled)}")

# 4. Build a simple Feedforward Neural Network model using Keras
model = keras.Sequential([
    keras.layers.Dense(units=10, activation='relu', input_shape=(X_train_nn_scaled.shape[1],)), # Hidden layer with 10 neurons and ReLU activation
    keras.layers.Dense(units=1, activation='sigmoid') # Output layer with 1 neuron and Sigmoid activation for binary classification
])

# 5. Compile the model
# We use 'adam' optimizer, 'binary_crossentropy' loss for binary classification, and 'accuracy' as a metric
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Display model summary
print("\nModel Summary:")
model.summary()

# 6. Train the model
print("\nTraining the Neural Network...")
history = model.fit(X_train_nn_scaled, y_train_nn, epochs=50, batch_size=32, verbose=0, validation_split=0.2)

# 7. Evaluate the model on the test set
loss, accuracy = model.evaluate(X_test_nn_scaled, y_test_nn, verbose=0)
print(f"\nNeural Network Test Accuracy: {accuracy:.4f}")

# Plot training history
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# Visualize decision boundary (for 2D data)
x_min, x_max = X_nn[:, 0].min() - 1, X_nn[:, 0].max() + 1
y_min, y_max = X_nn[:, 1].min() - 1, X_nn[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.05),
                     np.arange(y_min, y_max, 0.05))

# Predict probabilities for the mesh grid points
Z_proba = model.predict(scaler.transform(np.c_[xx.ravel(), yy.ravel()]), verbose=0)
Z = (Z_proba > 0.5).astype(int).reshape(xx.shape)

plt.figure(figsize=(8, 6))
plt.contourf(xx, yy, Z, cmap='bwr', alpha=0.5)
plt.scatter(X_nn[:, 0], X_nn[:, 1], c=y_nn, cmap='bwr', s=50, edgecolors='k')
plt.title('Neural Network Classifier Decision Boundary')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True)
plt.show()
```

## Types of Neural Networks

Neural networks come in various architectures, each designed for specific types of tasks and data:

1.  **Feedforward Neural Networks (FNNs) / Multi-Layer Perceptrons (MLPs)**:
    *   **Description**: The simplest type, where information flows in one direction from input to output, through hidden layers. Neurons in one layer are connected to all neurons in the next layer.
    *   **Use Cases**: Tabular data classification and regression, simple pattern recognition.
    *   **Example (already shown)**: The binary classification example we just explored was a simple FNN.

2.  **Convolutional Neural Networks (CNNs)**:
    *   **Description**: Specifically designed for processing data with a grid-like topology, such as images. They use convolutional layers to automatically and adaptively learn spatial hierarchies of features.
    *   **Use Cases**: Image classification, object detection, image segmentation, video analysis.

3.  **Recurrent Neural Networks (RNNs)**:
    *   **Description**: Designed to work with sequential data (like time series, text). They have internal memory, allowing them to retain information from previous inputs in the sequence.
    *   **Use Cases**: Natural Language Processing (NLP) tasks (e.g., language translation, text generation), speech recognition, time series prediction.

4.  **Long Short-Term Memory (LSTM) Networks**:
    *   **Description**: A special type of RNN capable of learning long-term dependencies. They use 'gates' to control the flow of information, mitigating the vanishing gradient problem common in vanilla RNNs.
    *   **Use Cases**: Similar to RNNs but often perform better on longer sequences.

5.  **Generative Adversarial Networks (GANs)**:
    *   **Description**: Consist of two neural networks, a generator and a discriminator, competing against each other to generate new, synthetic data that is indistinguishable from real data.
    *   **Use Cases**: Image generation, data augmentation, style transfer.


Let's start by delving into **Convolutional Neural Networks (CNNs)**, as they are fundamental for many computer vision tasks.

## Convolutional Neural Networks (CNNs)

Convolutional Neural Networks (CNNs) are a specialized type of neural network architecture designed to process pixel data in images. They are incredibly effective in tasks such as image recognition, object detection, and image segmentation. The key components of a CNN are:

1.  **Convolutional Layer**: This layer applies a set of learnable filters (kernels) to the input image, performing a convolution operation. Each filter detects specific features (edges, textures, patterns) in different regions of the image, producing a feature map.

2.  **Activation Function**: Typically ReLU (Rectified Linear Unit), applied after the convolutional layer to introduce non-linearity.

3.  **Pooling Layer (Subsampling)**: Reduces the spatial dimensions (width, height) of the feature maps, thereby reducing the number of parameters and computational cost, and making the network more robust to small variations in object position. Common pooling operations include Max Pooling and Average Pooling.

4.  **Fully Connected (Dense) Layer**: After several convolutional and pooling layers, the high-level features learned by these layers are flattened into a single vector and fed into one or more fully connected layers. These layers are similar to those in traditional Feedforward Neural Networks and perform the final classification or regression based on the extracted features.

Let's build a simple CNN to classify images from the **Fashion MNIST** dataset. This dataset contains 70,000 grayscale images of 10 fashion categories, with 60,000 for training and 10,000 for testing. Each image is 28x28 pixels.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
import numpy as np

# 1. Load the Fashion MNIST dataset
(train_images, train_labels), (test_images, test_labels) = keras.datasets.fashion_mnist.load_data()

# 2. Preprocess the data
# Reshape images to add a channel dimension (28, 28) -> (28, 28, 1)
train_images = train_images.reshape((60000, 28, 28, 1)).astype('float32') / 255
test_images = test_images.reshape((10000, 28, 28, 1)).astype('float32') / 255

# Verify shapes
print(f"Train images shape: {train_images.shape}")
print(f"Test images shape: {test_images.shape}")

# Define class names for plotting
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

# Display some example images
plt.figure(figsize=(10,10))
for i in range(25):
    plt.subplot(5,5,i+1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(train_images[i], cmap=plt.cm.binary)
    plt.xlabel(class_names[train_labels[i]])
plt.suptitle('Fashion MNIST Examples', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()

# 3. Build the CNN model
model_cnn = keras.Sequential([
    # Convolutional Layer 1
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),

    # Convolutional Layer 2
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Convolutional Layer 3 (optional, for deeper features)
    layers.Conv2D(128, (3, 3), activation='relu'),

    # Flatten the 3D output to 1D before feeding to Dense layers
    layers.Flatten(),

    # Fully Connected Layers
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax') # Output layer for 10 classes
])

# Display model summary
print("\nCNN Model Summary:")
model_cnn.summary()

# 4. Compile the model
model_cnn.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 5. Train the model
print("\nTraining the CNN...")
history_cnn = model_cnn.fit(train_images, train_labels, epochs=10, batch_size=64, verbose=1,
                            validation_data=(test_images, test_labels))

# 6. Evaluate the model
print("\nEvaluating the CNN on test data...")
test_loss_cnn, test_acc_cnn = model_cnn.evaluate(test_images, test_labels, verbose=0)
print(f"CNN Test Accuracy: {test_acc_cnn:.4f}")

# Plot training history
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(history_cnn.history['accuracy'], label='Training Accuracy')
plt.plot(history_cnn.history['val_accuracy'], label='Validation Accuracy')
plt.title('CNN Model Accuracy Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history_cnn.history['loss'], label='Training Loss')
plt.plot(history_cnn.history['val_loss'], label='Validation Loss')
plt.title('CNN Model Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
