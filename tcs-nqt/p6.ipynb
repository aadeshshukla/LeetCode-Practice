Question 1: Move All Zeros to the End (Array)Problem Statement:A chocolate factory packs chocolates into packets. Some packets are empty (represented by 0). Given an array of $N$ integers representing packet sizes, move all empty packets (0) to the end of the array while maintaining the relative order of the non-zero elements.  Input Format: An integer $N$ followed by $N$ space-separated array elements.Output Format: Space-separated integers with all zeros moved to the end.Example:Input: 8, [4, 5, 0, 1, 9, 0, 5, 0]Output: 4 5 1 9 5 0 0 0

```python
def move_zeros_to_end():
    n = int(input().strip())
    arr = list(map(int, input().split()))
    
    # Non-zero tracker pointer
    insert_pos = 0
    
    for i in range(n):
        if arr[i] != 0:
            arr[insert_pos] = arr[i]
            insert_pos += 1
            
    # Fill remaining spots with 0
    while insert_pos < n:
        arr[insert_pos] = 0
        insert_pos += 1
        
    print(*arr)

if __name__ == "__main__":
    move_zeros_to_end()
```

Question 2: Toggle Bits After Most Significant Bit (Bitwise Logic)Problem Statement:Given a positive integer $N$, convert it to its binary representation. Toggle all its bits starting from the Most Significant Bit (MSB) to the end, then convert the toggled binary back to a decimal integer and print it.  Example:Input: 10  Explanation: Binary of 10 is 1010. Toggling all bits yields 0101, which equals 5 in decimal.  Output: 5  

```python
import math

def toggle_bits():
    n = int(input().strip())
    
    # Calculate mask with all 1s up to the MSB position
    # E.g., for 10 (1010), bit length is 4 -> mask is 1111 (15)
    num_bits = int(math.log2(n)) + 1
    mask = (1 << num_bits) - 1
    
    # XOR with mask flips all bits
    result = n ^ mask
    print(result)

if __name__ == "__main__":
    toggle_bits()
```
