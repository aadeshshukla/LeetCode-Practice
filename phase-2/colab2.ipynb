```python
# Example DSA (Data Structures and Algorithms) problems

# 1. Reverse a string
def reverse_string(s):
    return s[::-1]

# 2. Find the maximum element in a list
def find_max(lst):
    return max(lst)

# 3. Check if a number is prime
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# 4. Fibonacci sequence up to n terms
def fibonacci(n):
    seq = [0, 1]
    for _ in range(2, n):
        seq.append(seq[-1] + seq[-2])
    return seq[:n]

# 5. Check if a string is a palindrome
def is_palindrome(s):
    return s == s[::-1]
```
